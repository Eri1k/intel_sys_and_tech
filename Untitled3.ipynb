```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score
import warnings
```

```python
#чтение данных
df = pd.read_excel('Тестовые данные айфон андроид.xlsx')
print(df.head())
```

```python
#удаление неинформативных столбцов (время и метро)
col_drop = ['Отметка времени', 'Ближайшая станция метро? (Точное название, например: Преображенская площадь, Сокольники и т.п.)']
df = df.drop(columns=[col for col in col_drop if col in df.columns])

#корректируем целевую переменную
target_col = 'У тебя айфон или андроид?'
df[target_col] = df[target_col].astype(str).str.strip().str.capitalize()
df = df[df[target_col].isin(['Айфон', 'Андроид'])]

#исправление ошибки
eye_col = 'Цвет глаз? (С большой буквы в именительном падеже единственное число, например: Зеленый)'
df = df[df[eye_col] != 'Кал']

#приаодим целевую переменную в числовой формат
df['target'] = df[target_col].apply(lambda x: 1 if x == 'Айфон' else 0)
df = df.drop(columns=[target_col])
#print(df.shape)
```

```python
#распределение на признаки
X = df.drop(columns=['target'])
y = df['target']

#дереление на исловые и категириальные
numeric_features = [
    'Как часто ты меняешь смартфон на новый? Введите число лет (целое, например: 2)',
    'Как часто вы играете в видеоигры? Введите примерное кол-во часов в неделю (целое, например: 5)'
]

categorical_features = [col for col in X.columns if col not in numeric_features]


for col in numeric_features:
    X[col] = pd.to_numeric(X[col], errors='coerce')
X = X.fillna(0)
```

```python
#перобразование данных
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ])

#создание модели
model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(n_estimators=100, random_state=42))
])
```

```python
#разбитие на выборки
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

#обучение самой модели
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

#оценка
print("\n--- Результаты ---")
print(f"Accuracy (Точность): {accuracy_score(y_test, y_pred):.2f}")
print("\nОтчет по классификации:")
print(classification_report(y_test, y_pred, target_names=['Андроид', 'Айфон']))
```
